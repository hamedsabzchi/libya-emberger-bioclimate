# Test Python connection to Google Earth Engine

This notebook checks only one thing: can Python connect to your Earth Engine account and read Libya + TerraClimate?

Important: do not upgrade Colab system packages here. Colab already includes the Earth Engine Python API.

```python
import ee
print('Earth Engine API version:', ee.__version__)
```

```python
PROJECT_ID = 'practical-proxy-441422-n6'

ee.Authenticate(auth_mode='notebook', force=True)
ee.Initialize(project=PROJECT_ID)

print('SUCCESS: Earth Engine authentication and initialization completed.')
```

```python
countries = ee.FeatureCollection('FAO/GAUL/2015/level0')
libya_fc = countries.filter(ee.Filter.eq('ADM0_NAME', 'Libya'))
libya = libya_fc.geometry()

country_name = libya_fc.first().get('ADM0_NAME').getInfo()
print('Boundary:', country_name)
```

```python
terraclimate = (
    ee.ImageCollection('IDAHO_EPSCOR/TERRACLIMATE')
    .filterDate('1991-01-01', '2021-01-01')
    .filterBounds(libya)
)

image_count = terraclimate.size().getInfo()
first_image = ee.Image(terraclimate.first())
first_date = ee.Date(first_image.get('system:time_start')).format('YYYY-MM-dd').getInfo()
bands = first_image.bandNames().getInfo()

print('SUCCESS: Python can access Google Earth Engine data.')
print('TerraClimate images for 1991-2020:', image_count)
print('First image date:', first_date)
print('Bands:', bands)
```
